# Data Loading

```python
batters = pd.read_csv('/kaggle/input/datasets/vinayaks0n1/ipl-2025/IPL2025Batters.csv')
bowlers = pd.read_csv('/kaggle/input/datasets/vinayaks0n1/ipl-2025/IPL2025Bowlers.csv')

batters.head()
```

```python
bowlers.head()
```

# Top Run Scorers

```python
top_runs = batters.sort_values(by='Runs', ascending=False).head(10)
top_runs[['Player Name', 'Runs']]
```

# Top Wicket Takers

```python
top_wickets = bowlers.sort_values(by='WKT', ascending=False).head(10)
top_wickets[['Player Name', 'WKT']]
```

# Top Six Hitters

```python
top_six = batters.sort_values(by='6s', ascending=False).head(10)
top_six[['Player Name', '6s']]
```

# Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.barplot(x=top_runs['Player Name'], y=top_runs['Runs'])
plt.xticks(rotation=45)
plt.title("Top Run Scorers IPL 2025")
plt.show()
```

# Final Insights

```python
print("\n--- Final Insights ---\n")
print("Top Run Scorer:", top_runs.iloc[0]['Player Name'], "-", top_runs.iloc[0]['Runs'], "runs")
print("Top Wicket Taker:", top_wickets.iloc[0]['Player Name'], "-", top_wickets.iloc[0]['WKT'], "wickets")
print("Top Six Hitter:", top_six.iloc[0]['Player Name'], "-", top_six.iloc[0]['6s'], "sixes")
```
